**2. Discount & Reward Functions**

```python
def calculate_discount(bill_amount, vip_member=False):


  if vip_member == True:
    total = bill_amount * 0.85

  else:
    total = bill_amount * 0.95

  return total


print("VIP Bill:", calculate_discount(1000, True))
print("Regular Bill:", calculate_discount(1000))
```

## Section 5: Object-oriented programming (OOP) Arichitecture

**1. Store Branch Class Hierarchy**

```python
class StoreBranch:

    company_name = "Urban Retail"

    def __init__(self, branch_id, daily_revenue):
        self.branch_id = branch_id
        self._daily_revenue = daily_revenue

    def get_revenue(self):
        return self._daily_revenue


class FlagshipStore(StoreBranch):

    def __init__(self, branch_id, daily_revenue, vip_lounge_access):
        super().__init__(branch_id, daily_revenue)
        self.vip_lounge_access = vip_lounge_access

    def get_revenue(self):
        return self._daily_revenue + 500.0


store = FlagshipStore(101, 2000.0, True)


print("Company Name:", StoreBranch.company_name)
print("Branch ID:", store.branch_id)
print("VIP Lounge Access:", store.vip_lounge_access)
print("Final Revenue:", store.get_revenue())
```
